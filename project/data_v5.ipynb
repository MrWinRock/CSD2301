```python
avm['Make_lower'] = avm['Make'].str.lower()
avm['Model_lower'] = avm['Model'].str.lower()
car['Make_lower'] = car['Make'].str.lower()
car['Model_lower'] = car['Model'].str.lower()

# Expand rows based on 'Unit' column
car_expanded = car.loc[car.index.repeat(car['Unit'])].reset_index(drop=True)

# Drop the 'Unit' column
car_expanded.drop(columns=['Unit'], inplace=True)

# Create a merged DataFrame
th_car_co2_emission = pd.merge(
    car_expanded,
    avm,
    how='inner',
    left_on=['Make_lower', 'Model_lower'],
    right_on=['Make_lower', 'Model_lower']
)

# Drop the temporary lowercase columns
th_car_co2_emission.drop(columns=['Make_lower', 'Model_lower'], inplace=True)

# Drop the duplicate Make and Model columns
th_car_co2_emission = th_car_co2_emission.loc[:, ~th_car_co2_emission.columns.duplicated()]

# Rename the columns to keep only one set of Make and Model
th_car_co2_emission.rename(columns={'Make_x': 'Make', 'Model_x': 'Model'}, inplace=True)
```

# Model

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Load your dataset into a DataFrame (assuming it's already loaded as 'th_car_co2_emission')
# th_car_co2_emission = pd.read_csv('your_dataset.csv')

# Print the column names to verify them
print(th_car_co2_emission.columns)

# Adjust the feature names based on the actual column names in your dataset
features = ['Engine size (L)', 'Cylinders', 'City (L/100 km)', 'Highway (L/100 km)', 'Combined (L/100 km)']
target = 'CO2 emissions (g/km)'

X = th_car_co2_emission[features]
y = th_car_co2_emission[target]

# Normalize the features using Min-Max scaling
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Define and train the KNN Regressor
knn_regressor = KNeighborsRegressor(n_neighbors=5)
knn_regressor.fit(X_train, y_train)
y_pred_knn = knn_regressor.predict(X_test)
mse_knn = mean_squared_error(y_test, y_pred_knn)
r2_knn = r2_score(y_test, y_pred_knn)
print(f'KNN Regressor MSE: {mse_knn:.4f}')
print(f'KNN Regressor R²: {r2_knn:.4f}')

# Define and train the Decision Tree Regressor
dt_regressor = DecisionTreeRegressor(random_state=42)
dt_regressor.fit(X_train, y_train)
y_pred_dt = dt_regressor.predict(X_test)
mse_dt = mean_squared_error(y_test, y_pred_dt)
r2_dt = r2_score(y_test, y_pred_dt)
print(f'Decision Tree Regressor MSE: {mse_dt:.4f}')
print(f'Decision Tree Regressor R²: {r2_dt:.4f}')

# Perform cross-validation for both models
cv_knn = cross_val_score(knn_regressor, X_scaled, y, cv=5, scoring='neg_mean_squared_error')
cv_dt = cross_val_score(dt_regressor, X_scaled, y, cv=5, scoring='neg_mean_squared_error')

print(f'KNN Regressor CV MSE: {-np.mean(cv_knn):.4f}')
print(f'Decision Tree Regressor CV MSE: {-np.mean(cv_dt):.4f}')
```

```python
plt.figure(figsize=(12, 6))

# Plot KNN predictions
plt.subplot(1, 2, 1)
plt.scatter(y_test, y_pred_knn, edgecolor='k', alpha=0.7, label='KNN Predictions')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2, label='Perfect Prediction Line')
plt.xlabel('Actual CO2 emissions (g/km)')
plt.ylabel('Predicted CO2 emissions (g/km)')
plt.title('KNN Regressor')
plt.legend(loc='upper left')

# Plot Decision Tree predictions
plt.subplot(1, 2, 2)
plt.scatter(y_test, y_pred_dt, edgecolor='k', alpha=0.7, label='Decision Tree Predictions')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2, label='Perfect Prediction Line')
plt.xlabel('Actual CO2 emissions (g/km)')
plt.ylabel('Predicted CO2 emissions (g/km)')
plt.title('Decision Tree Regressor')
plt.legend(loc='upper left')

plt.tight_layout()
plt.savefig('co2_emission_predictions.png')
plt.show()
```
